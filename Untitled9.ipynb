```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime, date
from dateutil.relativedelta import relativedelta

# -----------------------------
# Load data
# -----------------------------
CSV_PATH = "Nat_Gas.csv"
df = pd.read_csv(CSV_PATH)

# Parse dates (month-end snapshots)
df["Dates"] = pd.to_datetime(df["Dates"], format="%m/%d/%y")
df = df.sort_values("Dates").reset_index(drop=True)

# -----------------------------
# Seasonality + trend model (for extrapolation/backcast)
# Fit on log prices using linear regression with Fourier seasonality
# -----------------------------
y = np.log(df["Prices"].astype(float).values)

t0 = df["Dates"].min()
# time in years from start
t_years = (df["Dates"] - t0).dt.days.values / 365.25

# month of year as 1..12
m = df["Dates"].dt.month.values.astype(float)

def design_matrix(t_years, month, n_harmonics=2):
    """Build a simple seasonal-trend regression design matrix."""
    X = [np.ones_like(t_years), t_years]  # intercept + linear trend
    # Add Fourier seasonality terms
    for k in range(1, n_harmonics + 1):
        X.append(np.sin(2 * np.pi * k * month / 12.0))
        X.append(np.cos(2 * np.pi * k * month / 12.0))
    return np.column_stack(X)

X = design_matrix(t_years, m, n_harmonics=2)

# OLS fit: beta = (X'X)^-1 X'y
beta, *_ = np.linalg.lstsq(X, y, rcond=None)

def model_price(d: pd.Timestamp) -> float:
    """Model-implied price for any date (used for extrapolation/backcast)."""
    t = (d - t0).days / 365.25
    month = d.month
    Xd = design_matrix(np.array([t]), np.array([month]), n_harmonics=2)
    return float(np.exp(Xd @ beta)[0])

# -----------------------------
# Historical interpolation (for "any date in the past")
# Use linear interpolation between observed month-end snapshot prices.
# This gives an intuitive daily estimate inside the observed range.
# -----------------------------
obs_dates = df["Dates"].values.astype("datetime64[D]")
obs_days = obs_dates.astype("datetime64[D]").astype(int)
obs_prices = df["Prices"].astype(float).values

def estimate_gas_price(input_date) -> float:
    """
    Estimate natural gas purchase price for any date.
    - If input_date is inside observed range: linear interpolate between month-end points.
    - If outside observed range: use seasonal-trend model (extrapolation/backcast).

    input_date can be: 'YYYY-MM-DD', datetime, date, or pandas Timestamp.
    """
    if isinstance(input_date, str):
        d = pd.to_datetime(input_date)
    elif isinstance(input_date, datetime):
        d = pd.Timestamp(input_date)
    elif isinstance(input_date, date):
        d = pd.Timestamp(input_date)
    else:
        d = pd.Timestamp(input_date)

    d_day = np.datetime64(d.date(), "D").astype(int)
    if obs_days.min() <= d_day <= obs_days.max():
        # Linear interpolation in price levels
        return float(np.interp(d_day, obs_days, obs_prices))
    else:
        # Extrapolate/backcast using model
        return model_price(d)

# -----------------------------
# Extrapolate one extra year of month-end points
# -----------------------------
last_date = df["Dates"].max()
future_month_ends = []
d = last_date + relativedelta(months=1)
for _ in range(12):
    # ensure month-end by taking the last day of that month
    month_end = (pd.Timestamp(d.year, d.month, 1) + relativedelta(months=1)) - pd.Timedelta(days=1)
    future_month_ends.append(month_end)
    d = d + relativedelta(months=1)

future_df = pd.DataFrame({
    "Dates": future_month_ends,
    "Prices_Est": [model_price(d) for d in future_month_ends]
})

# -----------------------------
# Visualize: observed points + interpolated daily curve + 1y extrapolation
# -----------------------------
start = df["Dates"].min()
end = future_df["Dates"].max()

daily_dates = pd.date_range(start, end, freq="D")
daily_est = np.array([estimate_gas_price(d) if d <= last_date else model_price(d) for d in daily_dates])

plt.figure(figsize=(12, 5))
plt.plot(daily_dates, daily_est, label="Daily estimate (interp in-sample, model out-of-sample)")
plt.scatter(df["Dates"], df["Prices"], s=30, label="Observed month-end snapshots")
plt.scatter(future_df["Dates"], future_df["Prices_Est"], s=35, marker="x", label="1Y extrapolated month-ends")
plt.title("Natural Gas Month-End Snapshots + Daily Estimate + 1Y Extrapolation")
plt.xlabel("Date")
plt.ylabel("Price")
plt.legend()
plt.tight_layout()
plt.show()

# -----------------------------
# Example usage
# -----------------------------
examples = ["2021-02-15", "2022-12-15", "2024-09-30", "2025-03-15"]
for ex in examples:
    print(ex, "->", round(estimate_gas_price(ex), 4))

print("\nFuture month-end extrapolation (next 12 months):")
print(future_df.to_string(index=False))
```
